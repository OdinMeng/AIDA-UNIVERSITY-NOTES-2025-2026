```python
from typing import Generator
```

```python
def verify_prime(n: int) -> bool:
    if n < 2:
        return False 
    elif n == 2:
        return True 
    else:
        for i in range(2, int(n**0.5)+1):
            if n % i == 0:
                return False 
        else:
            return True

def generate_prime() -> Generator[int, None, None]:
    n = 2
    while True:
        if verify_prime(n):
            yield n
        n += 1
```

```python
gen = generate_prime()
```

```python
l = []
while len(l) != 101:
    num = next(gen)
    l.append(num)
```
